# Key Insights

1. Transaction amounts are highly skewed and contain substantial outliers.

2. Customer activity is uneven, with most customers making relatively few transactions.

3. Product and channel distributions are imbalanced, indicating potentially informative categorical variables.

4. Fraudulent transactions are rare, suggesting class imbalance in fraud-related analyses.

5. Customer transaction behavior supports the use of Recency, Frequency, and Monetary (RFM) metrics for proxy credit-risk labeling.

6. Transaction timing patterns suggest that temporal features may contribute useful behavioral information.
